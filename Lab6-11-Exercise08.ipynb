## Null strings and empty strings

Empty strings คือ object ของ string ที่ว่างเปล่า (ความยาวของข้อความที่เก็บใน string เป็นศูนย์) เราสามารถใช้งาน method ต่างๆ ของ string ได้ตามปกติ

Null strings คือ reference ที่ไม่ได้ชี้ไปยัง string ใดๆ เลย ถ้าเราพยายามใช้งาน จะเกิดการผิดพลาดที่ชื่อว่า  [NullReferenceException](https://learn.microsoft.com/en-us/dotnet/api/system.nullreferenceexception?view=net-8.0)  

อย่างไรก็ตาม เราสามารถใช้ null strings ในการเปรียบเทียบหรือเชื่อมต่อกับ string อื่นๆ

3. อธิบายโปรแกรม

ให้ศึกษาโปรแกรมด้านบนและอธิบายการทำงาน

1. การสร้าง String: สร้าง string ปกติ (str = "hello"), null string (nullStr = null), และ string ว่าง (emptyStr = String.Empty).
2. การเชื่อมต่อ String กับ null:
การเชื่อม nullStr (ที่เป็น null) กับ str จะให้ผลลัพธ์เป็น "hello" เนื่องจาก null ถูกแปลงเป็น string ว่าง

3. การเปรียบเทียบ emptyStr กับ nullStr:
emptyStr (string ว่าง) และ nullStr (null) ไม่เท่ากัน ผลลัพธ์คือ false

4. การสร้าง string ใหม่จาก emptyStr และ nullStr:
การเชื่อม emptyStr (string ว่าง) กับ nullStr จะให้ผลลัพธ์เป็น string ว่าง ("") ซึ่งมีความยาว 0

5. ข้อผิดพลาดจากการใช้ nullStr.Length:
การเข้าถึง .Length ของ null string จะเกิดข้อผิดพลาด NullReferenceException

6. การใช้ อักขระ null (\x0):
การเชื่อมต่อ string กับอักขระ null (Unicode \x0) จะทำให้ null นับเป็นอักขระหนึ่งตัวใน string
ความยาวของ string ที่เชื่อมกับ null จะเท่ากับจำนวนอักขระที่มี (เช่น 4 สำหรับ "abc" + "\x0")
